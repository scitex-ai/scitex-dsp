# 02 — Shape utilities: `ensure_3d`, `crop`

Two thin helpers for getting signals into the shape `scitex_dsp`
expects.

```python
import numpy as np
import scitex_dsp as dsp
%matplotlib inline

x1 = np.random.randn(1024)
x2 = np.random.randn(8, 1024)
x3 = np.random.randn(2, 8, 1024)
for x in (x1, x2, x3):
    print(f"{x.shape} -> {dsp.ensure_3d(x).shape}")
```

## `crop` — sliding-window segmentation

```python
xx, tt, fs = dsp.demo_sig(sig_type="chirp", batch_size=1, n_chs=1,
                          t_sec=4, fs=512)
seg = dsp.crop(xx, window_length=512, overlap_factor=0.5)
print("input :", xx.shape)
print("output:", np.asarray(seg).shape)
```
