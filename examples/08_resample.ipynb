# 08 — `resample`

`dsp.resample(x, src_fs, tgt_fs)` polyphase resampling.

```python

import numpy as np
import matplotlib.pyplot as plt
import scitex_dsp as dsp
%matplotlib inline

xx, tt, fs = dsp.demo_sig(sig_type="periodic", batch_size=1, n_chs=1,
                          t_sec=1, fs=1000, freqs_hz=[10])
yy = np.asarray(dsp.resample(xx, src_fs=fs, tgt_fs=250))
print("orig:", xx.shape, "@", fs, "Hz")
print("down:", yy.shape, "@ 250 Hz")

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(tt, xx[0, 0], label="1000 Hz")
ax.plot(np.linspace(tt[0], tt[-1], yy.shape[-1]), yy[0, 0],
        "o-", label="250 Hz", alpha=0.7)
ax.legend(); ax.set_xlabel("Time [s]"); ax.set_xlim(0, 0.3)
fig.tight_layout()
plt.show()
```
