# 03 — `norm`: per-trace normalization

- `dsp.norm.z(x, dim=-1)` — zero mean, unit std.
- `dsp.norm.minmax(x, amp=1.0, dim=-1)` — scale to ±amp.

Both apply along the last (time) axis by default.

```python

import numpy as np
import matplotlib.pyplot as plt
import scitex_dsp as dsp
%matplotlib inline

xx, tt, fs = dsp.demo_sig(sig_type="periodic", batch_size=1,
                          n_chs=1, t_sec=1, fs=512, freqs_hz=[5])
xx = xx * 50 + 100
xz  = np.asarray(dsp.norm.z(xx))
xmm = np.asarray(dsp.norm.minmax(xx))

fig, ax = plt.subplots(3, 1, figsize=(8, 5), sharex=True)
ax[0].plot(tt, xx[0, 0]);   ax[0].set_ylabel("orig")
ax[1].plot(tt, xz[0, 0]);   ax[1].set_ylabel("z-score")
ax[2].plot(tt, xmm[0, 0]);  ax[2].set_ylabel("minmax")
ax[-1].set_xlabel("Time [s]"); fig.tight_layout()
print("z mean,std:", float(xz.mean()), float(xz.std()))
plt.show()
```
